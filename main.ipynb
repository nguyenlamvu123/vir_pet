```python
import ipywidgets as widgets
from IPython.display import display
```

```python
class VirtualPet:
    def __init__(self, name):
        self.name = name
        self.hunger = 50
        self.happiness = 50
        self.energy = 50

    def feed(self, b=None):
        self.hunger = max(0, self.hunger - 10)
        self.happiness += 5
        status.value = f"{self.name} đang ăn 🍖. Đói: {self.hunger}, Vui vẻ: {self.happiness}"

    def play(self, b=None):
        self.happiness += 10
        self.energy -= 5
        status.value = f"{self.name} đang chơi 🎾. Vui vẻ: {self.happiness}, Năng lượng: {self.energy}"

    def sleep(self, b=None):
        self.energy += 15
        self.hunger += 5
        status.value = f"{self.name} đang ngủ 😴. Năng lượng: {self.energy}, Đói: {self.hunger}"
```

```python
pet = VirtualPet("Mèo Con")

btn_feed = widgets.Button(description="Feed")
btn_play = widgets.Button(description="Play")
btn_sleep = widgets.Button(description="Sleep")
status = widgets.Label(value="Thú cưng sẵn sàng 🐾")

btn_feed.on_click(pet.feed)
btn_play.on_click(pet.play)
btn_sleep.on_click(pet.sleep)

ui = widgets.VBox([btn_feed, btn_play, btn_sleep, status])
display(ui)
```
